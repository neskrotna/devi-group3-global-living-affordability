# Install project dependencies

Each teammate can use their own **Python 3.11–3.13 kernel**. This notebook installs `requirements.txt` into the Python kernel selected for this notebook.

1. Click **Run All** and wait for **Setup complete**. Internet access is needed for packages that are not installed yet.
2. Open [data_preprocessing.ipynb](data_preprocessing.ipynb) and select the **same Python kernel**.
3. **Restart Kernel**, then **Run All** in preprocessing.

Run setup once per Python environment, or again when `requirements.txt` changes. Packages that already satisfy the requirements are reused.

```python
from pathlib import Path
import subprocess
import sys

PROJECT_ROOT = next(
    (p for p in [Path.cwd(), *Path.cwd().parents]
     if (p / 'install_dependencies.py').is_file() and (p / 'requirements.txt').is_file()),
    None,
)
if PROJECT_ROOT is None:
    raise RuntimeError('Open this notebook from inside your copy of the project folder.')

print(f'Selected kernel Python: {sys.executable}', flush=True)
print('Installing and checking requirements. Please wait...', flush=True)
result = subprocess.run(
    [sys.executable, str(PROJECT_ROOT / 'install_dependencies.py')],
    cwd=PROJECT_ROOT,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    errors='replace',
)
print(result.stdout)
result.check_returncode()
```
