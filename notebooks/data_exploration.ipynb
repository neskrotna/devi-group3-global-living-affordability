### Dataset Version Selection

Two versions of the Global Cost of Living dataset were compared before further analysis.

The original dataset contains 4,874 records covering 4,816 unique cities and 204 countries. Version 2 contains 4,956 records covering 4,898 unique cities and 215 countries.

Neither dataset contains duplicate records. The original dataset also contains an additional `Unnamed: 0` column, which represents an unnecessary index column. Version 2 provides broader geographical coverage, with 96 locations not present in the original version.

Based on these observations, `cost-of-living_v2.csv` was selected for further analysis. Both original source files remain unchanged in the raw data directory.

# Data Exploration Conclusion

The initial exploration showed that all three datasets contain useful information for the Global Living Affordability analysis.

The **Cost of Living dataset** provides the most detailed information, with data for 4,956 records across 215 countries. It contains information about everyday expenses, rent, property prices, salaries, and other living costs. However, some variables have many missing values, most records have a low `data_quality` indicator, and some potential outliers were identified.

The **Income dataset** contains average annual wage data for 34 individual countries, plus one OECD aggregate. The dataset is complete and does not contain missing or duplicate values. The wages refer to 2025 and are expressed in constant US dollars using purchasing power parities (PPP).

The **Real Estate dataset** contains property price information for 62 countries for 2026. It does not contain missing or duplicate values, and the number of listings used to calculate the property prices is also available. Unlike the Cost of Living dataset, the property data is only available at country level.

Overall, the datasets use different geographical levels, country naming conventions, reference years, and monetary measures. Before they can be combined, the country names need to be standardized and the city-level Cost of Living data needs to be prepared for comparison with the country-level datasets. The different years and the use of PPP-adjusted income also need to be considered when interpreting the final results.
